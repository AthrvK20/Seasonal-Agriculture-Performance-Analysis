#9. Outlier Analysis

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []
for col in numeric_columns:
  q1 = df[col].quantile(0.25)
  q3= df[col].quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr
  count =((df[col] < lower) | (df[col] > upper)).sum()

  outlier_summary. append ({
    "Column": col,
    "Lower_Bound": lower,
    "Upper_Bound": upper,
    "Outlier_Count": count,
    "Outlier_Percentage": round(count / len(df) * 100, 2)
})
outlier_summary = pd.DataFrame(outlier_summary).sort_values(
  "Outlier_Count", ascending=False
)

outlier_summary
```

### 13.2 Crop Profitability and Consistency

```python
# 13.2 Crop Profitability and Consistency

crop_profitability = df.groupby('Crop')['Profit_INR'].agg(
    Total_Records='count',
    Profitable_Records=lambda x: (x > 0).sum()
)

crop_profitability['Profitability_Rate_%'] = (
    crop_profitability['Profitable_Records']
    / crop_profitability['Total_Records']
    * 100
)

crop_profitability = crop_profitability.sort_values(
    'Profitability_Rate_%',
    ascending=False
)

print("Crop Profitability Analysis:")
display(crop_profitability)
```

```python
# Profitability Rate by Crop

crop_profitability['Profitability_Rate_%'].plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Profitability Rate by Crop')
plt.xlabel('Crop')
plt.ylabel('Profitable Records (%)')
plt.xticks(rotation=45)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

# **14. Key Insights**


***Insight 1*: Agricultural performance varies considerably across seasons**

**Observation**:
Agricultural performance differs across Kharif, Rabi and Zaid seasons in terms of yield, profit and water usage.

**Evidence**:
Kharif has the highest average yield (5.63 tonnes/ha) and average profit (₹178,914.65). Rabi has an average yield of 5.04 tonnes/ha and average profit of ₹87,689.47. Zaid has the lowest average yield (4.64 tonnes/ha) and a negative average profit (−₹24,804.82). Zaid also has the highest average water usage at 6,419.89 m³.

**Interpretation**:
The results show substantial differences in agricultural and economic performance across seasons. Kharif shows the strongest overall performance in the available data, while Zaid shows comparatively weaker economic performance and higher average water usage.

**Limitation**:
The analysis is based on the available dataset and does not establish that season alone causes the differences. Crop selection, farming practices, environmental conditions and other factors may also contribute.

***Insight 2*: Environmental conditions differ substantially between seasons**

**Observation**:
The environmental conditions recorded in the dataset vary considerably across seasons.

**Evidence**:
Average rainfall is 849.20 mm in Kharif, 437.62 mm in Rabi and 304.65 mm in Zaid. Average temperature is 28.45°C, 23.49°C and 31.04°C, respectively. Average humidity is 71.81%, 57.89% and 52.01%.

**Interpretation**:
Each season is associated with a distinct combination of rainfall, temperature and humidity. These differences provide important context when comparing agricultural performance between seasons.

**Limitation**:
The notebook does not establish a direct causal relationship between these environmental conditions and agricultural outcomes. In particular, the rainfall-yield analysis shows only a very weak overall linear correlation (0.03).

***Insight 3*: Crop performance varies greatly, with Sugarcane showing exceptionally high performance**

**Observation**:
There are substantial differences in yield, profit and water efficiency among crops.

**Evidence**:
Across the seasonal crop comparison, Sugarcane has much higher average yield than the other crops. Its average yield is 53.46 tonnes/ha in Kharif, 43.29 tonnes/ha in Rabi and 38.42 tonnes/ha in Zaid. Its average profit is also substantially higher than the other crops in each season.

**Interpretation**:
Sugarcane is the strongest-performing crop in the dataset based on the analysed yield and profit measures. The results also show that crop choice is an important dimension when evaluating agricultural performance.

**Limitation**:
The crops have different production characteristics, and the dataset does not provide enough evidence to conclude that one crop is universally better than another outside the conditions represented in the dataset.

***Insight 4*: Production, revenue and profit show strong relationships with agricultural output**

**Observation**:
Several production and economic variables show meaningful positive relationships.

**Evidence**:
The correlation matrix shows a strong positive correlation between yield and production (0.88) and between revenue and profit (0.89). Farm area and production have a moderate positive correlation (0.58), while yield and profit have a moderate positive correlation (0.49).

**Interpretation**:
Higher yield is strongly associated with higher total production, while higher revenue is strongly associated with higher profit. However, yield alone does not completely determine profitability.

**Limitation**:
Correlation indicates association rather than causation. The relationships may also be influenced by other variables such as farm area, market price and production costs.

***Insight 5*: The dataset contains substantial outliers and highly skewed performance variables**

**Observation**:
Yield, production, water efficiency and profit contain a considerable number of potential outliers.

**Evidence**:
The IQR-based analysis identifies 404 potential outliers in Profit, 333 in Production, 322 in Water Efficiency, and 302 in Yield. The difference between mean and median is also substantial for some variables; for example, yield has a mean of 5.24 tonnes/ha but a median of only 1.74 tonnes/ha, while profit has a mean of ₹111,556.46 and median of ₹3,673.

**Interpretation**:
The large difference between mean and median and the presence of many potential outliers indicate that some observations have unusually high values and strongly influence the overall distributions. This explains why average performance should be interpreted together with distributions and median values.

**Limitation**:
The notebook identifies these observations as potential outliers using the IQR method but does not remove them. Therefore, some averages and correlations may be influenced by extreme observations.

# 15. Recommendations:

1. **Consider seasonal performance**: Use differences in yield, profitability, environmental conditions and water usage when planning agricultural activities.

2. **Choose irrigation methods carefully**: Evaluate irrigation methods along with the season, as profitability varies across different irrigation methods and seasons.

3. **Consider crop profitability**: Evaluate both yield and profitability consistency when selecting crops for cultivation.

4. **Improve water management**: Pay closer attention to water usage and efficiency, particularly during seasons with lower water-use efficiency.

5. **Monitor production costs and revenue**: Consider both production costs and revenue when evaluating agricultural performance, as higher yield does not always result in higher profit.

6. **Use multiple factors for decision-making**: Combine crop, seasonal, environmental, irrigation and economic information rather than relying on a single performance indicator.
